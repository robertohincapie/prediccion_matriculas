```python
from pathlib import Path
import pandas as pd
from matriculas.data import load_raw_data

RAW_DIR = Path("../data/raw")

dataset_path = (
    RAW_DIR
    / "reporte1_31_08_26"
    / "reporte1_31_08_26.xlsx"
)

df = load_raw_data(dataset_path)

df.head()
```

```python
#Vamos a verificar los tipos de datos que se tienen y la cantidad de no nulos que hay
df.info()
```

```python
#En principio que soberva que el período es un int, pero no tiene sentido. Hay que convertirlo a 
#semestre y año. Para eso vamos a crear una nueva columna que se llame "periodo" y que contenga el año y el semestre.
df['period'].unique()
#Se puede ver que hay valores 21, 41, 50, etc en los semestres que por ahora eliminaré. Solo me 
#quedaré con valores 10 y 20. Para eso voy a crear una nueva columna que se llame "periodo" y que contenga el año y el semestre.
```

```python
from matriculas.prepare import prepare_period

df_periods = prepare_period(df, column="period")

df_periods[
    ["period", "year", "semester", "time_index"]
].drop_duplicates().sort_values("time_index")
```
